Step 2: Data Cleaning & Preprocessing

```
# Check dataset size
print("Dataset Shape:", df.shape)

# Display column names
print("\nColumn Names:")
print(df.columns.tolist())

# Check missing values
print("\nMissing Values:")
print(df.isnull().sum())

# Check duplicate records
print("\nDuplicate Records:", df.duplicated().sum())
```

Step 3: Data Formatting and Type Conversion

```

# Step 3: Data Formatting and Type Conversion

import pandas as pd
import numpy as np

# Check data types before conversion
print("Before Conversion:")
print(df.dtypes)

# Remove extra spaces from column names
df.columns = df.columns.str.strip()

# Convert numeric columns
numeric_columns = [
    'Price', 'MPG.city', 'MPG.highway',
    'EngineSize', 'Horsepower', 'RPM',
    'Rev.per.mile', 'Fuel.tank.capacity',
    'Passengers', 'Length', 'Wheelbase', 'Width', 'Weight'
]

for col in numeric_columns:
    if col in df.columns:
        df[col] = pd.to_numeric(df[col], errors='coerce')

# Clean text columns
for col in df.select_dtypes(include=['object', 'string']).columns:
    df[col] = df[col].astype('string').str.strip()

# Check data types after conversion
print("\nAfter Conversion:")
print(df.dtypes)

print("\nCars93 data formatting completed successfully!")

# Display first five rows
display(df.head())
```

Step 5: EDA – Data Analysis

```



import pandas as pd
import numpy as np

# Display first five records
print("First 5 Records:")
display(df.head())

# Display dataset shape
print("\nDataset Shape:")
print("Rows:", df.shape[0])
print("Columns:", df.shape[1])

# Display column names
print("\nColumn Names:")
print(df.columns.tolist())

# Statistical summary
print("\nStatistical Summary:")
display(df.describe())

# Check unique car manufacturers
if 'Manufacturer' in df.columns:
    print("\nNumber of Manufacturers:")
    print(df['Manufacturer'].nunique())

# Average price
if 'Price' in df.columns:
    print("\nAverage Car Price:")
    print(round(df['Price'].mean(), 2))

# Highest horsepower
if 'Horsepower' in df.columns:
    print("\nMaximum Horsepower:")
    print(df['Horsepower'].max())

# Average city mileage
if 'MPG.city' in df.columns:
    print("\nAverage City Mileage:")
    print(round(df['MPG.city'].mean(), 2))

print("\nEDA completed successfully!")
```
